### Linguagem orientada a objetos

Tudo dentro da programação Python é um objeto, e para que seja possível realizar análises desses objetos devemos ser capazes de acessar esses objetos.

Logo, todo objeto deve ter sua etiqueta.

```python
# utilizando o Python como calculadora
((100-80)/80)*100
```

```python
# criando um objeto
variacao_percentual = ((100-80)/80)*100
```

```python
# acessando o objeto diretamente
variacao_percentual
```

```python
# acessando o objeto pela função print
print(variacao_percentual)
```

```python
# acessando o objeto pela função print
print('O preço do produto cresceu variacao_percentual% entre junho e julho')
```

```python
# acessando o objeto pela função print
print('O preço do produto cresceu {}% entre junho e julho'.format(variacao_percentual))
```

```python
# criando um objeto de texto
frase = 'O preço do produto cresceu {}% entre junho e julho'
```

```python
# combinando objetos
print(frase.format(variacao_percentual))
```

```python
# criando novo objeto de texto
frase = 'O preço do produto cresceu {}% entre julho e agosto'
```

```python
# combinando objetos
print(frase.format(variacao_percentual))
```

A frase antiga *'O preço do produto cresceu {}% entre junho e julho'* virou **lixo**. Logo, ela é descartada pelo Python e o espaço ocupado por ela se torna disponível novamente.
