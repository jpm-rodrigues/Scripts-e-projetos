# Função any
A função any() é usada em Pandas para verificar se pelo menos um valor True está presente em um conjunto de valores. Ela pode ser aplicada a colunas, linhas ou a um DataFrame inteiro.
A estrutura geral é:

```python
df.any(axis=X)
```

axis=0 (ou omitido) → Verifica colunas, ou seja, se pelo menos um valor True existe em cada coluna.
axis=1 → Verifica linhas, ou seja, se pelo menos um valor True existe em cada linha.
skipna=True (padrão) → Ignora valores NaN na verificação.

# Criando e manipulando listas


```python
lista = [4, 5, 6]
lista
# Adicionar coisas ao final da lista
lista.append(1)
lista.append(2)
lista
# Remover coisas no final da lista
lista.pop()
lista
# Acessar a lista com o índice
lista[1]
```

     # Sintaxe de slice

```python
# Lembrando que o último número do index nunca é incluido
lista[1:3]  # Pega os elementos de 1 até 3, sem incluir o três
lista[2:]   # Pega os elementos de 2 até o final
lista[:3]   # Pega os elementos do início até o índice 3
lista[::2]  # Após o primeiro :, o segundo faz steps, ou seja, vai pulando de tantos em tantos elementos
lista[::-1] # Retorna a lista em reverso
# A sintaxe é basicamente essa
# lista[start:end:step]
```
